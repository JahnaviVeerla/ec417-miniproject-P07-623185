```python
import sys
from pathlib import Path
project_root = Path.cwd().resolve()
if project_root.name == 'notebooks':
    project_root = project_root.parent
sys.path.insert(0, str(project_root))

import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import seaborn as sns

from src.data import FEATURE_COLUMNS, load_wine_dataset, summarize_duplicates
```

```python
data_dir = project_root / 'data' / 'raw'
df = load_wine_dataset(data_dir)
print('Rows:', len(df))
print(df[['quality','is_red','good']].head())
```

```python
before, after, cleaned = summarize_duplicates(df)
print(f'Duplicates before: {before}')
print(f'Duplicates after: {after}')
print('Good ratio before:', df['good'].mean())
print('Good ratio after:', cleaned['good'].mean())
```

```python
sns.histplot(df['quality'], bins=10, kde=False)
plt.title('Wine quality histogram')
plt.xlabel('quality')
plt.ylabel('count')
plt.tight_layout()
plt.savefig(project_root / 'results' / 'figures' / 'eda_quality_histogram.png', dpi=200)
plt.close()
```

```python
sns.boxplot(data=df, x='is_red', y='alcohol', hue='is_red')
plt.title('Alcohol by wine type')
plt.xlabel('is_red (1=red, 0=white)')
plt.ylabel('alcohol')
plt.tight_layout()
plt.savefig(project_root / 'results' / 'figures' / 'eda_alcohol_type.png', dpi=200)
plt.close()
```

```python
corr = df[FEATURE_COLUMNS + ['quality']].corr()
fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(corr, annot=False, cmap='coolwarm', ax=ax)
plt.title('Feature correlation heatmap')
plt.tight_layout()
plt.savefig(project_root / 'results' / 'figures' / 'eda_correlation_heatmap.png', dpi=200)
plt.close()
```

```python
sns.boxplot(data=df, x='good', y='density', hue='is_red')
plt.title('Density by class and color')
plt.xlabel('good')
plt.ylabel('density')
plt.tight_layout()
plt.savefig(project_root / 'results' / 'figures' / 'eda_density_boxplot.png', dpi=200)
plt.close()
```

```python
print('EDA note: the dataset is strongly skewed toward white wines and class imbalance is present; density and alcohol are highly associated, as expected in wine chemistry. Duplicates are numerous and were removed before splitting.')
```
